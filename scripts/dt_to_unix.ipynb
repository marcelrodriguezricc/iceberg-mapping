```python
# Converts datetime to unix time so point clouds output from QPS Qimera can be processed in Cloud Compare

# To use, modify in_dir to point at directory containing Qimera outputs and out_dir to desired location for saving modified point cloud tables


import pandas as pd
from pathlib import Path
in_dir = Path("/Users/marcel/Desktop/iceberg-motion/data/2024-06-17/cleaned/datetime")
out_dir = Path("/Users/marcel/Desktop/iceberg-motion/data/2024-06-17/cleaned/unix_time")
out_dir.mkdir(parents=True, exist_ok=True)
for i, txt_path in enumerate(sorted(in_dir.glob("*.txt"))):
    df = pd.read_csv(txt_path)
    df.columns = df.columns.str.replace("#", "", regex=False).str.replace(" ", "", regex=False)
    df["DateTime"] = pd.to_datetime(df["DateTime"]).astype("int64") / 10**9
    df = df.rename(columns={"DateTime": "UnixTime"})
    df.to_csv(out_dir / f"pass_{i:02d}.csv", index=False)
```
