```python
# FILE 0

import pandas as pd
from pathlib import Path

IN_DIR = Path("../data/2024-06-14/file_point_clouds")
OUT_DIR = Path("../data/2024-06-14/pass_point_clouds")
OUT_DIR.mkdir(parents=True, exist_ok=True)

input_file = IN_DIR / "file_00.csv"

cutpoint = 1718393975.70655823
cutpoint1 = 1718394053.11184263
cutpoint2 = 1718394164.94990730
cutpoint3 = 1718394262.57661152
cutpoint4 = 1718394315.71869612

df = pd.read_csv(input_file)

pass1 = df[df["UnixTime"] < cutpoint]
pass2 = df[(df["UnixTime"] >= cutpoint) & (df["UnixTime"] <= cutpoint1)]
pass3 = df[(df["UnixTime"] >= cutpoint1) & (df["UnixTime"] <= cutpoint2)]
pass4 = df[(df["UnixTime"] >= cutpoint2) & (df["UnixTime"] <= cutpoint3)]
pass5 = df[(df["UnixTime"] >= cutpoint3) & (df["UnixTime"] <= cutpoint4)]
pass6 = df[df["UnixTime"] >= cutpoint4]

pass1.to_csv(OUT_DIR / "file_00_pass_00.csv", index=False)
pass2.to_csv(OUT_DIR / "file_00_pass_01.csv", index=False)
pass3.to_csv(OUT_DIR / "file_00_pass_02.csv", index=False)
pass4.to_csv(OUT_DIR / "file_00_pass_04.csv", index=False)
pass5.to_csv(OUT_DIR / "file_00_pass_05.csv", index=False)
```
