```python
import pandas as pd
from pathlib import Path
from datetime import datetime, timezone

IN_DIR = Path("../data/2024-06-14/pass_point_clouds")

for f in sorted(IN_DIR.glob("*.csv")):
    t = pd.read_csv(f, usecols=["UnixTime"])["UnixTime"]
    t_start = datetime.fromtimestamp(t.min(), tz=timezone.utc)
    t_end   = datetime.fromtimestamp(t.max(), tz=timezone.utc)
    print(f"{f.name:<30}  {t_start:%Y-%m-%d %H:%M:%S}  →  {t_end:%Y-%m-%d %H:%M:%S} UTC")
```
